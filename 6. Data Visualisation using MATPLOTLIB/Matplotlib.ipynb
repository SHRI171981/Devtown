Background

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x=np.array([10,11,22,13,14])
y=np.array([20,18,21,15,14])
```

```python
plt.plot(y,'*')
# line joined graph
```

Types of lines:

```python
plt.plot(y,'o')
# 'o' -> Markers
```

```python
plt.plot(y,':')
# dotted line
```

```python
plt.plot(y,'--')
# dashed line
```

```python
plt.plot(y,'-.')
# line joined grap
```

Format String

```python
plt.plot(y,'*--b',ms=25,mec='k',mfc='w',color='r',linewidth=4)

# MARKUP-LINE-COLOR convention:

# * -> point
# -- -> line
# b -> line color
 

# Using parameters: 

# ms -> marker size
# mec -> marker edge color
# mfc -> marker face color
# color -> line color
# linewidth -> line size

```

Two lines in one plot

```python
plt.plot(x)
plt.plot(y)
```

Label, Title etc


```python
plt.plot(y)
plt.plot(x)
plt.xlabel('Level')
plt.ylabel('Difficulty',loc='top')
modification={'color':'b','size':22,'family':'calibri'}
plt.title('Level - Difficulty graph',fontdict=modification,loc='left')


plt.show()
```

-> GRID

```python
plt.plot(y)
plt.plot(x)
plt.grid(axis='y',color='r',linewidth='2',ls='--')
#Customizations same as line
plt.show()
```

SUBPLOT

```python
#plt.subplot(rows,columns,plot_number)

#PLOT 1
x1=[1,2,3,4,5,6]
y1=[10,20,30,40,50,60]
plt.subplot(2,2,1)
plt.plot(x1,y1)
plt.title('P1')

#PLOT 2
x2=np.array([10,11,22,13,14])
y2=np.array([20,18,21,15,14])
plt.subplot(2,2,2)
plt.plot(x2,y2)
plt.title('P2')

#PLOT 3
x3=np.array([14,12,12])
y3=np.array([20,18,21])
plt.subplot(2,2,3)
plt.plot(x3,y3)
plt.title('P3')

#PLOT 4
x4=np.array([22,13,14])
y4=np.array([20,15,14])
plt.subplot(2,2,4)
plt.plot(x4,y4)
plt.title('P4')

plt.suptitle('Super Title')
```

TICKS and TICKS Labels

```python
x=[1,2,3,4,5,6]
y=[10,20,30,40,50,60]

fig=plt.figure()
ax=fig.add_axes([1,1,2.5,1.5])
ax.set_title('Animal')
ax.set_xticks([0.5,1,1.5,2,2.5,3,3.5,4,4.5,5,5.5,6,6.5,7]) #tick values changed
ax.set_yticklabels([1,2,3,4,5,6]) #only tick label changed
ax.plot(x,y)
```

Bar-Plot

```python
x=['A','B','C','D']
y=[12,8,4,10]
plt.subplot(1,2,1)
plt.bar(x,y)
plt.subplot(1,2,2)
plt.barh(x,y)
```

Histogram

Pie-Chart

```python
x=[15,25,21,10]
plt.pie(x,labels=['A','B','C','D'],explode=[0.5,0.1,0,0.25],counterclock=True,shadow=True,colors=['red','blue','white','yellow'])
plt.legend(title='Labels:')
```
