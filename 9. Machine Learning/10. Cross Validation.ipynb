Machine Learning

```python
X=(df.iloc[:,2:])
Y=(df.iloc[:,1])
```

```python
Y.isnull().sum()
```

Using train_test_split

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=15)

model=LogisticRegression()

model.fit(X_train,Y_train)

result=model.score(X_test,Y_test)
print(result)

#Random State = 0 -> 0.9473684210526315
#Random State = 100 -> 0.956140350877193
#Random State = 500 -> 0.9385964912280702
#Random State = 15 -> 0.8947368421052632

del result,model
```

## Cross-Validation

1. K Fold CV

```python
from sklearn.model_selection import KFold,cross_val_score
```

```python
k = KFold(10)
```

```python
model=LogisticRegression()
result=cross_val_score(model,X,Y,cv=10)
print(result)
print(np.mean(result))
```

2. Stratified KFold CV

```python
from sklearn.model_selection import StratifiedKFold
```

```python
skfold=StratifiedKFold(n_splits=5)
result=cross_val_score(model,X,Y,cv=skfold)
print(np.mean(result))
```

3. LOO CV

```python
from sklearn.model_selection import LeaveOneOut
```

```python
LOO=LeaveOneOut()
result=cross_val_score(model,X,Y,cv=LOO)
print(np.mean(result))
#result->0.945518453427065
```
