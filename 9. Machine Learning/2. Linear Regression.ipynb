 # Linear Regression

### Simple Linear Regression

importing the libs

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

given data

```python
exp=np.array([1.1,1.3,1.5,2.0,2.2,2.9,3.0,3.2,3.2,3.7,3.9,4.0,4.0,4.1,4.5,4.9,5.1,5.3,5.9,6.0,6.1,7.9,8.2,8.7,9.0,9.5,9.6,10.3,10.5,12])
sal=np.array([39343,46205,37731,43535,39821,56642,60150,54445,64445,57189,63218,55794,56957,57081,61111,67938,83088,81363,93940,91738,98217,101302,113812,109431,105582,116906,126350,122391,121872,143000])
```

```python
X=exp.reshape(-1,1)
Y=sal
```

```python
plt.scatter(exp,sal)
```

Assigning dataframe to the dataset

```python
df=pd.DataFrame({"Exp":exp,"Salary":sal})
```

```python
df.info()
```

- x -> Feature
- Y -> Target

Installing sklearn library for ML

importing the package for dividing data

```python
from sklearn.model_selection import train_test_split
```

splitting the data

    - Train data = 70%
    - Test data = 30%

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=15,shuffle=True)
```

Choosing the Model -> Linear Reg model

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

Fitting the model

```python
model.fit(X_train,Y_train)
```

Make Predictions

```python
Y_test #expected output
```

```python
Y_pred=model.predict(X_test) #predicted output
```

Performance Metrics

```python
from sklearn.metrics import*
```

MSE

```python
mean_squared_error(Y_test,Y_pred)
```

Accuracy

```python
model.score(X_test,Y_test)*100
```

R-squared

```python
r2_score(Y_test,Y_pred)*100
```

Predict Value

```python
model.predict([[9.5]])[0]
```
