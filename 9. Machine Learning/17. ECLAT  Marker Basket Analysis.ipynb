ECLAT

```python
from pyECLAT import ECLAT
```

```python
eclat=ECLAT(data=df,verbose=True)
```

```python
eclat.df_bin
```

```python
eclat_indexes,eclat_support=eclat.fit_all(min_support=0.03,separator=' -> ',verbose=True,min_combination=2)
```
