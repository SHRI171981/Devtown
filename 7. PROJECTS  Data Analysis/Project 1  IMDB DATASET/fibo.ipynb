```python
memo={}
def Fibo(n):
    if n<=1:
        value=n
        memo[n]=value
        return value
    try:
        value1=memo[n-1]
    except KeyError:
        value1=Fibo(n-1)
        memo[n-1]=value1
    try:
        value2=memo[n-2]
    except KeyError:
        value2=Fibo(n-2 )
        memo[n-2]=value2
    
    value=value1+value2
    return value

    
```

```python
Fibo(6)
```

```python
for i in range(101):
    print('Fibo',i,':',Fibo(i))
   
```
