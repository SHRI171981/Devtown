```python
word=input('Enter a word')
word=word.lower()
n=len(word)
ans=[]
for i in range(n):
    ans+='_'
no_of_misses=7
```

```python
Guesses=[]
Misses=[]
j=1
print(f'Guesses:{Guesses}')
print(f'Misses:\t{Misses}')
print(f'Answer->',end='\t')
for x in ans:
            print(x,end=' ')
print('\n')
while '_' in ans and len(Misses)<no_of_misses:
    inp=input('Guess a letter!')
    print('Try',j,':',inp)
    if inp in word:
        Guesses.append(inp)
        for k in range(n):
            if word[k]==inp:
                ans[k]=inp
        print(f'Guesses:{Guesses}')
        print(f'Misses:\t{Misses}')
        print(f'Answer->\t',end='\t')
        for x in ans:
            print(x,end=' ')
        print('\n')
    else:
        Misses.append(inp)
        print(f'Guesses:{Guesses}')
        print(f'Misses:\t{Misses}')
        print(f'Answer\t->',end='\t')
        for x in ans:
            print(x,end=' ')
        print('\n')
    j+=1
ans=[]
j=0

if '_' in ans:
     print("Sorry you lost")
else:
     print("Congratulations!! You guessed the right word.")
    
```
